```python

def greet():
    print("Hello! Welcome to Python Functions.")

greet()
```

```python

def greet(name):
    print(f"Hello, {name}!")

greet("saniya")
greet("rii")
```

```python
def greet(name="Guest"):
    print(f"Welcome, {name}!")

greet()
greet("saniya")
```

```python
def print_names(*names):
    print("Names passed:", names)
    for name in names:
        print("Hello", name)

print_names("hoi", "rii", "aa")
```

```python
def print_data(**info):
    print("User data:")
    for key, value in info.items():
        print(f"{key} = {value}")

print_data(name="saniya", age=20, city="hyd")
```

```python
def show_details(*args, **kwargs):
    print("Args:", args)
    print("Kwargs:", kwargs)

show_details(10, 20, 30, name="Rii", city="hyd")
```

```python
def multiply(a, b):
    return a * b

def square(n):
    return multiply(n, n)

print("Square of 5:", square(5))
```
