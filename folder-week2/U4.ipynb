```python
a = np.array([1, 2, 3])
b = np.array([4, 0, 1])
print('a . b  (dot)   :', np.dot(a, b))
print('a x b  (cross) :', np.cross(a, b))
```

```python
print('L2 norm          :', la.norm(a))
print('L1 norm          :', la.norm(a, 3))
print('Linf norm        :', la.norm(a, np.inf))
def cosine(u, v):
    return np.dot(u, v) / (la.norm(u) * la.norm(v))

print('cosine(a, b)     :', round(cosine(a, b), 3))
```

```python
pairs = [
    (np.array([1, 0, 1]), np.array([1, 0, 1])),
    (np.array([1, 2, 3]), np.array([3, 2, 1])),
    (np.array([2, 0, 0]), np.array([0, 5, 0])),
]

cosine_similarities = []
max_similarity = -1
most_similar_pair_idx = -1

for i, (u, v) in enumerate(pairs, start=1):
    print(f"\nPair {i}:")
    l1_norm_u = la.norm(u, 1)
    l2_norm_u = la.norm(u)
    print(f"  L1 norm of u: {l1_norm_u:.3f}")
    print(f"  L2 norm of u: {l2_norm_u:.3f}")
    cos_sim = cosine(u, v)
    print(f"  Cosine similarity of u and v: {cos_sim:.3f}")
    cosine_similarities.append(cos_sim)
    if cos_sim > max_similarity:
        max_similarity = cos_sim
        most_similar_pair_idx = i
print(f"\nMost similar pair is Pair {most_similar_pair_idx} with a cosine similarity of {max_similarity:.3f}")
```

```python
A = np.array([[2., 1.],
              [1., 3.]])
B = np.array([[1., 0.],
              [4., 2.]])

print('A @ B (matrix multiply):\n', A @ B)
print('A.T   (transpose):\n', A.T)
print('inverse(A):\n', la.inv(A))
print('trace(A) = sum of diagonal:', np.trace(A))
```

```python
I = np.eye(2)
print('Identity I:\n', I)
print('A @ inv(A) == I ?', np.allclose(A @ la.inv(A), I))


print('A symmetric?', np.allclose(A, A.T))

theta = np.radians(30)
Q = np.array([[np.cos(theta), -np.sin(theta)],
              [np.sin(theta),  np.cos(theta)]])
print('Q orthogonal?', np.allclose(Q.T @ Q, I))
```

```python

A = np.array([[2., 0.],
              [0., 3.]])

vals, vecs = la.eig(A)
print('Eigenvalues  (lambda):', vals)
print('Eigenvectors (columns):\n', vecs)
```

```python
for i in range(len(vals)):
    v = vecs[:, i]
    lhs = A @ v
    rhs = vals[i] * v
    print(f'lambda={vals[i]:.1f}  A v == lambda v ?', np.allclose(lhs, rhs))
```

```python
A = np.array([[ 2.,  1., -1.],
              [-3., -1.,  2.],
              [-2.,  1.,  2.]])
b = np.array([8., -11., -3.])

x = la.solve(A, b)
print('Solution x:', x)
print('Rank of A :', la.matrix_rank(A))
print('Full rank -> a unique solution exists')
```
